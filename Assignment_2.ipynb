```python
#1. Student Grade System
Marks < 35 → Fail
Marks ≥ 35 and < 75 → Grade C
Marks ≥ 75 and < 90 → Grade B
Marks ≥ 90 → Grade A
```

```python
print(f'{'*'*10}Student Grade System{'*'*10}')
marks = int(input('Enter the Student Grade:'))
if (marks >= 35):
    if(marks >= 75):
        if (marks >= 90):
            print('Grade A')
        else:
            print('Grade B')
    else:
        print('Grade C')
else: 
    print('fail')
```

```python
#2. Number Check

Number divisible by 2 → Even
If even and divisible by 4 → Special Even
Number not divisible by 2 → Odd
If odd and divisible by 3 → Special Odd
```

```python
print(f'{'*'*10}Number Check{'*'*10}')
num = int(input('Enter a number:'))
if (num%2!=0):
    if (num%2!=0) and (num/3):
        if(num%2==0 and num/4== 0):
            print('Special Even')
        else:
            print('Even')
    else: 
        print('Special Odd')
else:
    print('Even')
```

```python
# 3. Age Classification

Age < 18 → Minor
Age ≥ 18 and < 60 → Adult
Age ≥ 60 → Senior Citizen
```

```python
print(f'{'*'*10}Age Classification{'*'*10}')
age = int(input('Enter the age:'))
if (age >= 18):
    if (age >= 60):
        print('Senior Citizen')
    else:
        print('Adult')
else:
    print('Minor')
```

```python
#4. Temperature Category

Temp < 15 → Cold
Temp ≥ 15 and ≤ 30 → Moderate
Temp > 30 and ≤ 40 → Hot
Temp > 40 → Very Hot
```

```python
print(f'{'*'*10}Temperature Category{'*'*10}')
temp = int(input('Enter the temperature:'))
if (temp >= 15):
    if(temp > 30):
        if(temp > 40):
            print('Very Hot')
        else:
            print('Hot')
    else:
        print('Moderate')
else:
    print('cold')
```
