```python
#1. Student Grade System
''' 
Marks < 35 → Fail
Marks ≥ 35 and < 75 → Grade C
Marks ≥ 75 and < 90 → Grade B
Marks ≥ 90 → Grade A 
'''
```

```python
print(f'{'*'*10}Student Grade System{'*'*10}')
marks = int(input('Enter the Student Grade:'))
if (marks >= 35):
    if(marks >= 75):
        if (marks >= 90):
            print('Grade A')
        else:
            print('Grade B')
    else:
        print('Grade C')
else: 
    print('fail')
```

```python
#2. Number Check
'''
Number divisible by 2 → Even
If even and divisible by 4 → Special Even
Number not divisible by 2 → Odd
If odd and divisible by 3 → Special Odd
'''
```

```python
print(f'{'*'*10}Number Check{'*'*10}')
num = int(input('Enter a number:'))
if (num%2!=0):
    if (num%2!=0) and (num/3):
        if(num%2==0 and num/4== 0):
            print('Special Even')
        else:
            print('Even')
    else: 
        print('Special Odd')
else:
    print('Even')
```

```python
#7. Electricity Bill
'''
Units ≤ 50 → ₹2/unit
Units > 50 and ≤ 100 → ₹3/unit
Units > 100 and ≤ 200 → ₹5/unit
Units > 200 → ₹8/unit
'''
```

```python
print(f'{'*'*10}Electricity Bill{'*'*10}')
units = float(input('Enter the units:'))
if (units > 50):
    if(units > 100):
        if(units > 200):
            print(f'Payable amount is {(50*2)+(200*5)+((units-250)*8)}')
        else:
            print(f'Payable amount is {(50*2)+(100*3)+((units-150)*5)}')
    else:
        print(f'Payable amount is {(50*2)+((units - 50)*3)}')
else:
    print(f'Payable amount is {(units*2)}')
```

```python
#8. Shopping Discount
'''
Amount < 500 → No discount
Amount ≥ 500 and < 1000 → 5% discount
Amount ≥ 1000 and < 5000 → 10% discount
Amount ≥ 5000 → 20% discount
'''
```

```python
#9. Loan Eligibility
'''
Salary ≥ 25000 and credit score ≥ 750 → Approved
Salary ≥ 25000 and credit score < 750 → Review
Salary < 25000 and credit score ≥ 800 → Approved
Otherwise → Rejected
'''
```

```python
print(f'{'*'*10}Loan Eligibility{'*'*10}')
sal = float(input('Enter the Salary:'))
cs = float(input('Enter the Credit Score:'))
if(sal >= 25000):
    if(cs < 750):
        print('Review')
    else:
        print('Approved')
elif (sal < 25000):
    if(cs >= 800):
        print('Approved')
else:
    print('Rejected')
print(f'{'*'*30}')
```

```python
#10. Exam Result System
'''
All subjects ≥ 35 → Pass
Any subject < 35 → Fail
If pass:
Average ≥ 85 → Distinction
Average ≥ 60 and < 85 → First Class
Average < 60 → Pass Class
If any subject < 20 → Fail (Severe)
'''
```

```python
print(f'{'*'*10}Exam Result System{'*'*10}')
sub1 = float(input('Enter marks of English:'))
if (sub1 > 100):
    print('Invalid')
sub2 = float(input('Enter marks of kannada:'))
if (sub2 > 100):
    print('Invalid')
sub3 = float(input('Enter marks of Math:'))
if (sub3 > 100):
    print('Invalid')
sub4 = float(input('Enter marks of science:'))
if (sub4 > 100):
    print('Invalid')
sub5 = float(input('Enter marks of Social:'))
if (sub5 > 100):
    print('Invalid')

allsub = sub1+sub2+sub3+sub4+sub5
if (sub1<35 or sub2<35 or sub3<35 or sub4<35 or sub5<35):
    print('Fail')
elif (allsub>= 35):
    print('Pass')
    avg = allsub/5
    if (avg<60):
        if(avg<85):
            print('Pass Class')
        else:
            print('Distinction')
    else:
        print('First Class')
elif (sub1<20 or sub2<20 or sub3<20 or sub4<20 or sub5<20):
    print('Fail (Severe)')
print(f'{'*'*45}')
```

```python
#11. Ride Fare Calculation
'''
Distance ≤ 5 km and non-peak → ₹10/km
Distance ≤ 5 km and peak → ₹15/km
Distance > 5 km and non-peak → ₹8/km
Distance > 5 km and peak → ₹12/km
'''
```

```python
print(f'{'*'*10}Ride Fare Calculation{'*'*10}')
dis = float(input('Enter the Distance in Km:'))
pe = str(input('Enter the time period:'))
if (pe == 'peak' or 'non-peak'):
    if(dis>5):
        if(pe=='peak'):
            print('₹12/km')
            fare =dis*12
            print('Ride Fare:',fare)
        if (pe=='non-peak'):
            print('₹8/km')
            fare =dis*8
            print('Ride Fare:',fare)
    elif(dis<=5):
        if(pe=='peak'):
            print('₹15/km')
            fare =dis*15
            print('Ride Fare:',fare)
        if(pe=='non-peak'):
            print('₹10/km')
            fare =dis*10
            print('Ride Fare:',fare)
else:
    print('Enter whether peak or non-peak')
print(f'{'*'*50}')
```
